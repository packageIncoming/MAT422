```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
rng = np.random.default_rng(422)
```

## Joint probability distributions

This section gives one compact Python demonstration of **Joint probability distributions**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
joint = np.array([[0.05, 0.20, 0.10],
                  [0.10, 0.25, 0.30]])
x_values = np.array([0, 1])
y_values = np.array([0, 1, 2])

px = joint.sum(axis=1)
py = joint.sum(axis=0)
print("Joint total =", joint.sum())
print("P(X) =", px)
print("P(Y) =", py)
print("P(Y=2 | X=1) =", joint[1, 2] / px[1])
```

- Joint distributions are probabilities for two or more events at once. In this example it's probabilities of the variables X and Y (both discrete) where X={0,1} and Y= {0,1,2}.
- The sum of all probabilities in the joint table is 1. Summing down columns give P(Y), and summing across rows gives P(X).


## Correlation and dependence

This section gives one compact Python demonstration of **Correlation and dependence**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
x = rng.normal(0, 1, 200)
y_dependent = 2 * x + rng.normal(0, .5, 200)
y_independent = rng.normal(0, 3, 200)

print("corr(x, dependent y) =", np.corrcoef(x, y_dependent)[0, 1])
print("corr(x, independent y) =", np.corrcoef(x, y_independent)[0, 1])

plt.scatter(x, y_dependent, alpha=0.7, label="dependent")
plt.scatter(x, y_independent, alpha=0.5, label="mostly independent")
plt.legend()
plt.title("Correlation as a dependence summary")
plt.show()
```

- Two events/variables are correlated if they have some sort of relationship. Here it might mean that they occur in some sort of predictable pattern.
- In this example we're given two variables, y_independent and y_dependent, and we compare the correlation to the random distribution (x).
- Since y_dependent is a function of that same random distribution, it is more correlated.
- But y_independent is an entirely different random distribution and therefore not correlated.
The correlation coefficients show this too, as the correlation coefficient of y_dependent is far higher than y_independent.


## Random samples

This section gives one compact Python demonstration of **Random samples**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
sizes ={1,5,10,30}
for s in sizes:
  sample_means = []
  for _ in range(1000):
      sample = rng.exponential(scale=2.0, size=s)
      sample_means.append(sample.mean())

  sample_means = np.array(sample_means)
  print("Mean of sample means =", sample_means.mean())
  print("Std of sample means =", sample_means.std())

  plt.hist(sample_means, bins=30, color="tab:green", edgecolor="white")
  plt.title("Sampling distribution of the sample mean")
  plt.xlabel("sample mean")
  plt.ylabel("count")
  plt.show()
```

This section shows the central limit theorem in action. It takes a sample of 1000 means from exponential distributions and graphs the means. What's interesting is that the exponential distribution is heavily skewed, but as the size of the sample increases, the distribution becomes a bell-shaped distribution instead.
