## Probability axioms

This section gives one compact Python demonstration of **Probability axioms**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
outcomes = np.array([2, 2, 3, 4, 5, 6])
probabilities = np.ones(6) / 6
event_a = outcomes % 2 == 0
event_b = outcomes >= 5

print("Probabilities nonnegative:", np.all(probabilities >= 0))
print("Total probability:", probabilities.sum())
print("P(A: even) =", probabilities[event_a].sum())
print("P(B: at least 5) =", probabilities[event_b].sum())
print("P(A union B) =", probabilities[event_a | event_b].sum())
print("P(A) + P(B) - P(A and B) =", probabilities[event_a].sum() + probabilities[event_b].sum() - probabilities[event_a & event_b].sum())
```

Interpretations:
1. Probability is always positive
2. Probability always sums to 1
3. P(A)U P(B) = P(A) + P(B) - P(A and B)
  -You're effectively removing the intersection so that it's not double-counted.

This example specifically focuses on the sides of an even die.

## Conditional probability

This section gives one compact Python demonstration of **Conditional probability**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
prevalence = 0.08
sensitivity = 0.90
false_positive = 0.07

p_positive = sensitivity * prevalence + false_positive * (1 - prevalence)
p_disease_given_positive = sensitivity * prevalence / p_positive

print("P(positive) =", round(p_positive, 4))
print("P(disease | positive) =", round(p_disease_given_positive, 4))
```

This specific example is on the probability of a positive test (on a patient having a disease) is a **true positive** rather than a false positive.

P(positive) represents the probability of an individual from the population getting a positive indication of having the disease (whether it is a false or true positive).

P(disease | positive) (read as the probability of having the disease given the test being positive) represents the odds of a positive testing being a **true positive** reading. Basically, if you're given a positive reading, what're the odds that you really have the disease?

## Discrete random variables

This section gives one compact Python demonstration of **Discrete random variables**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
x = np.array([1, 1, 2, 3])
pmf = np.array([0.10, 0.25, 0.40, 0.25])
expected_value = np.sum(x * pmf)
variance = np.sum((x - expected_value) ** 2 * pmf)

print("E[X] =", expected_value)
print("Var(X) =", variance)

plt.bar(x, pmf, color="tab:blue")
plt.xlabel("x")
plt.ylabel("P(X=x)")
plt.title("Discrete random variable PMF")
plt.show()
```

Interpretation:
Discrete random variables are sets of a finite number of elements each with their own probability. The mapping of element to probability is the Probability Mass Function.

The expected value for a DRV is Σ(a*pmf(a)) for all a in the DRV.

Informally, the variance represents how spread out the data is (higher-> more spread out).

## Continuous random variables

This section gives one compact Python demonstration of **Continuous random variables**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
grid = np.linspace(0, 8, 500)
rate = 1
pdf = rate * np.exp(-rate * grid)
approx_area = np.trapezoid(pdf, grid)
p_less_than_2 = np.trapezoid(pdf[grid <= 2], grid[grid <= 2])

print("Approximate total area under PDF =", approx_area)
print("Approximate P(X <= 2) =", p_less_than_2)

plt.plot(grid, pdf, color="tab:orange")
plt.fill_between(grid[grid <= 2], pdf[grid <= 2], alpha=0.3)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Continuous random variable: exponential PDF")
plt.show()
```

Interpretation:

Continuous random variables are random variables whose potential values are continuous and therefore countably infinite. Instead of fininite values like [0,1,2,3,4], it might be something instead like all numbers between [0,100].

Still, CRVs obey probability rules. The sum of all probabilities must be 1. However, the probability mass function does not apply to CRVs, as the probability of a single value is 0. Instead CRVs use probability **density** funcitons. There can be different distributions, this example for instance uses an exponential distribution.

Probabilities are given in terms of ranges or slices rather than specific numbers, integrating from (start) to (end).
